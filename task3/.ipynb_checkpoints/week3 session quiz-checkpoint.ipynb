```python
import math as m
import scipy.constants as scp

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']

vel={}

for i in range(len(masses)):
    velocity = m.sqrt(masses[i]*2*scp.G/radii[i])
    vel[names[i]]= velocity
    print(f"{masses[i]:.2e}")
print(vel)
```

```python
import math as m
def tangentthing(d,D):
    o=m.degrees(2*m.atan(d/D/2))
    return o
print(tangentthing(1,2))
```

```python
import scipy.constants as scp
def cyllinder(r,h):
    A= 2*scp.pi*r**2 +2*scp.pi*r*h
    V= scp.pi*r**2*h
    return A,V
area,volume = cyllinder(5,7) 
ratio=area/volume
print(f"{ratio}")
```

```python

import numpy as np
a= np.linspace(1,10,10)
def stdev(a):
    mu = np.sum(a)/len(a)
    o = np.sqrt(np.sum((a- mu)**2)/len(a))
    return o

def stdev2(a):
    o = np.sqrt(np.sum((a-(np.sum(a)/len(a)))**2)/len(a))
    return o
    
x1=stdev(a)
x2=stdev2(a)
print(a)
print(x1)
print(x2)
print(np.std(a))
```
