```python
def summary(*array):
    ret=np.zeros(len(array[1]))
    for i in array:
        for x in range(len(i)):
            ret[x]=ret[x]+i[x]
    
    print(ret)

a=np.array([5,4,3,2,1])
b=np.array([1,2,3,4,5])
summary(a,b)
```

```python
from astropy.cosmology import Planck18
import scipy as scp
import math
import numpy as np

H0 = Planck18.H0.value
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0
H0=H0*1000/(3.085677581*10**22)

def f(x):
    hz=H0*math.sqrt(Om0*(1+x)**3+Ode0)
    func=1/((1+x)*hz)
    return func

print(f"{np.array(scp.integrate.quad(f,0,2))/(10**9*365*24*3600)} Gyr")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---- Generate example data ----

x = np.random.normal(0, 1, 100000)
y = np.random.normal(2, 1.0, 100000)

# ---- Create 2D histogram ----
H, xedges, yedges,image = plt.hist2d(x, y, bins=30,cmap='Blues')
plt.colorbar()

# transpose so axes match matplotlib
H = H/H.sum()
Hflat=H.flatten()
Hsort=np.sort(Hflat)[::-1]

cdf=np.cumsum(Hsort)
levels=[Hsort[np.searchsorted(cdf,p)] for p in [0.68,0.95,0.997]]
levels=sorted(levels)


# ---- Create grid for contour plotting ----
xcenters = (xedges[:-1] + xedges[1:]) / 2
ycenters = (yedges[:-1] + yedges[1:]) / 2
X, Y = np.meshgrid(xcenters, ycenters)

# ---- Plot ----


contours=plt.contour(
    X, Y, H.T,
    levels=levels,
    colors="red",
    linewidths=2
)


plt.gca().set_aspect(1)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(ls='dashed',color='black')
plt.title("2D Histogram with σ Contours")
plt.clabel(contours,fmt={levels[2]:'1std',levels[1]:'2std',levels[0]:'3std'})


plt.show()
```
