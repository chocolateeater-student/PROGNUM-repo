```python
from scipy.constants import c

class Galaxy:    
    """Class for redshift calculation of galaxies"""  
        
    caH_0 = 3968.5      # Calcium H line in rest frame
    
    def __init__(self, name, calciumHline=caH_0,c=c):
        self.name = name
        self.caH = calciumHline
        self.c = c
    def redshift(self):        
        return (self.caH - self.caH_0)/self.caH
    def velocity(self):
        z = self.redshift()
        v=z*self.c/1000
        return v

gal = Galaxy("NGC1832", 3994.7)
print(gal.velocity())
```

```python
import numpy as np
from matplotlib.pyplot import figure, show
from matplotlib.patches import Ellipse
import matplotlib.pyplot as plt


fig = figure()
frame = fig.add_subplot(1,1,1, aspect=1)

Nx = 400
Ny = 300
data = np.random.random((Ny,Nx))
frame.imshow(data,interpolation='none', origin="lower", cmap='gray')

xcenter, ycenter = 200,150
width = 200; height= 100
angle = 30.0
# Note that the angle parameter is a keyword parameter
e1 = Ellipse((xcenter, ycenter), width, height, angle=angle, fill=False)
frame.add_patch(e1)

frame.grid()
show()
```
